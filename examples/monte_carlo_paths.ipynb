# Monte Carlo price paths

This notebook can be used to generate and plot underlying asset price paths using Monte Carlo simulation, assuming geometric Brownian motion and risk neutrality.

```python
from optionlab import __version__
from optionlab.mc import create_price_path_bs, plot_price_paths
import matplotlib.pyplot as plt
import sys
```

```python
print(f"Python version: {sys.version}")
print(f"OptionLab version: {__version__}")
```

## Input

You must provide the initial underlying asset price, the annualized volatility and continuously compounded risk-free interest rate (both expressed as decimals), the number of trading days, the number of time intervals, and the number of paths. A fixed random seed makes the simulation reproducible.

In this example, 1,000 paths start at $100, with 20% annual volatility and a 5% annual risk-free interest rate. The simulation covers 252.0 trading days divided into 2,520 intervals.

```python
s0 = 100.0
volatility = 0.2
interest_rate = 0.05
days_to_maturity = 252.0
n_intervals = 2520
n_paths = 1000
seed = 42
```

## Calculations

The price paths are generated using `create_price_path_bs()`. The function converts trading days to years using 252 trading days per year, so this example spans one year with ten intervals per trading day.

The returned array contains one path per row and one time point per column, including the initial price. Its shape is therefore `(1000, 2521)`.

```python
paths = create_price_path_bs(
    s0=s0,
    days_to_maturity=days_to_maturity,
    volatility=volatility,
    interest_rate=interest_rate,
    n_intervals=n_intervals,
    n_paths=n_paths,
    seed=seed,
)
```

## Output

You can find below the plot of the simulated price paths.

```python
plot_price_paths(paths)
plt.show()
```
